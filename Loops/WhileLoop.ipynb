A practical beginner-friendly notebook covering `while` loop syntax, conditions, counters, user input, `break`, `continue`, `else`, nested loops, common patterns, and practice problems.

## 1. What is a `while` loop?

A `while` loop repeatedly executes a block of code **as long as its condition is `True`**. It is useful when you do not know in advance how many repetitions are needed.

````python
# Print a message 3 times
count = 1

while count <= 3:
    print("Hello, Python!")
    count += 1  # Update the counter
````

## 2. Basic syntax

```python
while condition:
    statements
```

1. Python checks the condition.
2. If it is `True`, the indented block runs.
3. Python checks the condition again.
4. The loop stops when the condition becomes `False`.

**Important:** If the condition never becomes false, the loop may run forever. This is called an infinite loop.

````python
number = 1

while number <= 5:
    print(number)
    number = number + 1

print("Loop finished")
````

## 3. Understanding the counter

A counter controls how many times a loop runs. Common updates are `count += 1`, `count -= 1`, and `count = count + 2`.

````python
# Count from 1 to 5
count = 1
while count <= 5:
    print(count)
    count += 1

print("---")

# Count backwards from 5 to 1
count = 5
while count >= 1:
    print(count)
    count -= 1
````

## 4. Using comparison and logical operators

A `while` condition can use operators such as `==`, `!=`, `<`, `>`, `<=`, `>=`, `and`, `or`, and `not`.

````python
# Continue while the number is between 1 and 5
number = 1
while number >= 1 and number <= 5:
    print(number)
    number += 1
````

## 5. `while` loop with `if` and `else`

````python
number = 1
while number <= 10:
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")
    number += 1
````

## 6. `break`: stop the loop

`break` immediately exits the nearest loop.

````python
number = 1
while number <= 10:
    if number == 5:
        break
    print(number)
    number += 1

# Output: 1, 2, 3, 4
````

## 7. `continue`: skip the current iteration

`continue` skips the remaining statements in the current iteration and starts the next check. Make sure the counter is updated before `continue` when necessary, otherwise you can create an infinite loop.

````python
number = 0
while number < 5:
    number += 1
    if number == 3:
        continue
    print(number)

# Output: 1, 2, 4, 5
````

## 8. `pass`: placeholder

`pass` does nothing. It can be used as a placeholder while planning a loop.

````python
count = 0
while count < 3:
    pass  # Placeholder only; do not run this as written!
    count += 1

print("Finished")
````

## 9. `while...else`

The `else` block runs when the loop condition becomes false. It does **not** run if the loop exits with `break`.

````python
count = 1
while count <= 3:
    print(count)
    count += 1
else:
    print("Loop completed normally")

print("---")

count = 1
while count <= 5:
    if count == 3:
        print("Stopping early")
        break
    print(count)
    count += 1
else:
    print("This will not print")
````

## 10. Taking user input with a `while` loop

A loop can keep asking for input until the user enters a particular value.

````python
# Keep asking until the user enters 'stop'
user_text = ""
while user_text.lower() != "stop":
    user_text = input("Enter something (type stop to finish): ")

print("Program ended")
````

## 11. Input validation

A `while` loop can repeat until the user provides valid input.

````python
age = int(input("Enter your age (1 to 120): "))

while age < 1 or age > 120:
    print("Invalid age. Try again.")
    age = int(input("Enter your age (1 to 120): "))

print("Your age is", age)
````

## 12. Nested `while` loops

A nested loop is a loop inside another loop. The inner loop completes its iterations for each outer-loop iteration.

````python
row = 1
while row <= 3:
    column = 1
    while column <= 3:
        print(row, column)
        column += 1
    row += 1
````

## 13. Pattern printing

````python
# Right-angled star pattern
row = 1
while row <= 5:
    print("*" * row)
    row += 1
````

## 14. Practical examples

````python
# A. Sum numbers from 1 to 10
number = 1
total = 0
while number <= 10:
    total += number
    number += 1
print("Sum:", total)

# B. Multiplication table
table_number = 5
multiplier = 1
while multiplier <= 10:
    print(f"{table_number} x {multiplier} = {table_number * multiplier}")
    multiplier += 1

# C. Count digits in a positive integer
number = 58392
digit_count = 0
while number > 0:
    number //= 10
    digit_count += 1
print("Digit count:", digit_count)
````

````python
# D. Reverse a positive integer
number = 12345
reversed_number = 0

while number > 0:
    last_digit = number % 10
    reversed_number = reversed_number * 10 + last_digit
    number //= 10

print("Reversed number:", reversed_number)

# E. Find the sum of digits
number = 4726
digit_sum = 0
while number > 0:
    digit_sum += number % 10
    number //= 10
print("Sum of digits:", digit_sum)
````

## 15. `for` loop vs `while` loop

| `for` loop | `while` loop |
|---|---|
| Often used to iterate over a sequence | Often used while a condition remains true |
| Common when items or repetitions are known | Common when repetitions depend on a condition or user input |
| The iterator usually advances automatically | You often update the condition variable yourself |

Both loops can solve many of the same problems. Choose the one that makes the logic clearer.

## 16. Common mistakes to avoid

1. **Forgetting to update the counter:** the condition may stay true forever.
2. **Wrong comparison:** check whether you need `<`, `<=`, `>`, or `>=`.
3. **Incorrect indentation:** the loop body must be indented.
4. **Updating in the wrong place:** especially when using `continue`.
5. **Not converting input:** `input()` returns a string. Use `int()` or `float()` when numeric input is required.
6. **Accidental infinite loops:** test the condition and counter carefully. Stop a running cell with the notebook's interrupt/stop control if needed.

## 18. Practice solutions

````python
# 1. Print 1 to 10
number = 1
while number <= 10:
    print(number)
    number += 1

# 2. Print 10 down to 1
number = 10
while number >= 1:
    print(number)
    number -= 1

# 3. Even numbers from 2 to 20
number = 2
while number <= 20:
    print(number)
    number += 2

# 4. Sum from 1 to 50
number = 1
total = 0
while number <= 50:
    total += number
    number += 1
print("Sum:", total)
````

````python
# 5. Multiplication table
table_number = int(input("Enter a number: "))
multiplier = 1
while multiplier <= 10:
    print(f"{table_number} x {multiplier} = {table_number * multiplier}")
    multiplier += 1

# 6. Password validation (example password: python123)
password = ""
while password != "python123":
    password = input("Enter password: ")
print("Access granted")
````

````python
# 7. Count digits (works for positive integers)
number = 58392
digits = 0
if number == 0:
    digits = 1
else:
    while number > 0:
        digits += 1
        number //= 10
print("Digits:", digits)

# 8. Sum of digits
number = 4726
total = 0
while number > 0:
    total += number % 10
    number //= 10
print("Digit sum:", total)

# 9. Star pattern
row = 1
while row <= 5:
    print("*" * row)
    row += 1

# 10. Stop at 7 using break
number = 1
while number <= 10:
    if number == 7:
        break
    print(number)
    number += 1
````

## 19. Quick revision

| Concept | Purpose |
|---|---|
| `while condition:` | Repeat while condition is true |
| Counter update | Changes the condition so the loop can end |
| `break` | Exits the loop immediately |
| `continue` | Skips to the next iteration |
| `pass` | Placeholder that does nothing |
| `else` | Runs when the loop ends normally, without `break` |
| Nested `while` | A `while` loop inside another `while` loop |

**Key lesson:** Always ask yourself: what makes this condition become `False`? If nothing changes it, you may have an infinite loop.
