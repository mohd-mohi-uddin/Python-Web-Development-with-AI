## 1. What is a `for` loop?

A `for` loop repeats a block of code for each item in an iterable (such as a list, string, tuple, dictionary, or range).

````python
# Example: print a message 3 times
for i in range(3):
    print("Hello, Python!")
````

## 2. Basic syntax

```python
for variable in sequence:
    statements
```

- `for` starts the loop.
- `variable` receives the current item on each iteration.
- `in` connects the variable to the sequence.
- The indented block runs once for each item.

````python
# The variable takes each value from the list, one at a time
colours = ["red", "blue", "green"]

for colour in colours:
    print(colour)
````

## 3. The `range()` function

`range()` produces a sequence of integers. Its stop value is **not included**.

- `range(stop)` starts at 0.
- `range(start, stop)` starts at `start`.
- `range(start, stop, step)` moves by `step` each time.

````python
# 0 to 4
for number in range(5):
    print(number)

print("---")

# 1 to 5
for number in range(1, 6):
    print(number)

print("---")

# Even numbers from 2 to 10
for number in range(2, 11, 2):
    print(number)
````

````python
# Count backwards from 5 to 1
for number in range(5, 0, -1):
    print(number)
````

## 4. Iterating over strings

A string is iterable. The loop visits each character in order.

````python
word = "Python"

for character in word:
    print(character)
````

## 5. Iterating over lists and tuples

````python
fruits = ["apple", "mango", "banana"]

for fruit in fruits:
    print("I like", fruit)

print("---")

scores = (75, 82, 91)
for score in scores:
    print(score)
````

## 6. Iterating over dictionaries

By default, a dictionary loop gives its keys. Use `.items()` to get key-value pairs.

````python
student = {"name": "Asim", "age": 22, "course": "Python"}

# Keys
for key in student:
    print(key)

print("---")

# Values
for value in student.values():
    print(value)

print("---")

# Keys and values
for key, value in student.items():
    print(key, "=", value)
````

## 7. Using `if` and `else` inside a `for` loop

````python
for number in range(1, 11):
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")
````

## 8. `break`: stop the loop

`break` immediately exits the nearest loop.

````python
for number in range(1, 10):
    if number == 5:
        break
    print(number)

# Output: 1, 2, 3, 4
````

## 9. `continue`: skip the current iteration

`continue` skips the remaining code in the current iteration and moves to the next one.

````python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)

# Output: 1, 2, 4, 5
````

## 10. `pass`: placeholder

`pass` does nothing. It is useful when you want to write a loop structure now and add its body later.

````python
for number in range(3):
    pass  # No action yet

print("The loop finished.")
````

## 11. Nested `for` loops

A nested loop is a loop inside another loop. For every outer-loop iteration, the inner loop runs completely.

````python
for row in range(1, 4):
    for column in range(1, 4):
        print(row, column)
    print("End of row", row)
````

## 12. Pattern printing with nested loops

````python
# Right-angled star pattern
for row in range(1, 6):
    for column in range(row):
        print("*", end="")
    print()  # Move to the next line
````

## 13. The `for...else` structure

The `else` block runs if the loop finishes normally. It does **not** run if the loop exits using `break`.

````python
for number in range(1, 4):
    print(number)
else:
    print("Loop completed normally")

print("---")

for number in range(1, 5):
    if number == 3:
        print("Found 3")
        break
else:
    print("This will not print")
````

## 14. Useful built-in functions: `enumerate()` and `zip()`

- `enumerate()` gives an index and an item.
- `zip()` combines items from multiple iterables position by position.

````python
names = ["Asha", "Ravi", "John"]
for index, name in enumerate(names, start=1):
    print(index, name)

print("---")

students = ["Asha", "Ravi", "John"]
marks = [88, 76, 91]
for name, mark in zip(students, marks):
    print(name, "scored", mark)
````

## 15. Common practical examples

````python
# A. Sum of numbers from 1 to 10
total = 0
for number in range(1, 11):
    total += number
print("Sum:", total)

# B. Multiplication table
table_number = 5
for number in range(1, 11):
    print(f"{table_number} x {number} = {table_number * number}")

# C. Count vowels in a string
text = "Learning Python"
vowels = "aeiouAEIOU"
count = 0
for character in text:
    if character in vowels:
        count += 1
print("Vowel count:", count)

# D. Find the largest number in a list
numbers = [12, 45, 7, 89, 23]
largest = numbers[0]
for number in numbers:
    if number > largest:
        largest = number
print("Largest:", largest)
````

## 16. Common mistakes to avoid

1. **Forgetting the colon:** write `for item in items:`.
2. **Incorrect indentation:** statements inside the loop must be indented consistently.
3. **Off-by-one errors:** `range(1, 6)` stops before 6.
4. **Wrong step direction:** use a negative step when counting backwards.
5. **Changing a list while iterating over it:** this can skip items. When removing items, iterate over a copy or build a new list.
6. **Confusing `break` and `continue`:** `break` exits the loop; `continue` skips one iteration.

## 18. Practice solutions

Run these only after attempting the questions.

````python
# 1. Numbers from 1 to 20
for number in range(1, 21):
    print(number)

# 2. Reverse order
for number in range(20, 0, -1):
    print(number)

# 3. Odd numbers from 1 to 50
for number in range(1, 51):
    if number % 2 != 0:
        print(number)

# 4. Sum from 1 to 100
total = 0
for number in range(1, 101):
    total += number
print(total)

# 5. Multiplication table from user input
table_number = int(input("Enter a number: "))
for number in range(1, 11):
    print(f"{table_number} x {number} = {table_number * number}")
````

````python
# 6. Count vowels
word = input("Enter a word: ")
vowel_count = 0
for character in word:
    if character.lower() in "aeiou":
        vowel_count += 1
print("Vowels:", vowel_count)

# 7. Count positive numbers
values = [-2, 5, 0, 8, -1, 3]
positive_count = 0
for value in values:
    if value > 0:
        positive_count += 1
print("Positive numbers:", positive_count)

# 8. Largest number without max()
values = [14, 62, 9, 37, 51]
largest = values[0]
for value in values:
    if value > largest:
        largest = value
print("Largest:", largest)

# 9. Star pattern
for row in range(1, 6):
    print("*" * row)

# 10. Names and marks
names = ["Asha", "Ravi", "John"]
marks = [88, 76, 91]
for name, mark in zip(names, marks):
    print(f"{name}: {mark}")
````
